```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python

img = cv2.imread('sample.jpg')

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python

gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.show()
```

```python

blur = cv2.GaussianBlur(gray, (5, 5), 0)

plt.imshow(blur, cmap='gray')
plt.title('Gaussian Blurred Image')
plt.axis('off')
plt.show()
```

```python

edges = cv2.Canny(blur, 50, 150)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title('Original Grayscale')

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')

plt.show()
```

```python

edge1 = cv2.Canny(blur, 30, 100)
edge2 = cv2.Canny(blur, 50, 150)
edge3 = cv2.Canny(blur, 100, 200)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(edge1, cmap='gray')
plt.title('Low Threshold (30, 100)')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(edge2, cmap='gray')
plt.title('Medium Threshold (50, 150)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(edge3, cmap='gray')
plt.title('High Threshold (100, 200)')
plt.axis('off')

plt.show()
```

```python

median = cv2.medianBlur(gray, 5)

median_edges = cv2.Canny(median, 50, 150)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(edges, cmap='gray')
plt.title('Gaussian Filter + Canny')

plt.subplot(1, 2, 2)
plt.imshow(median_edges, cmap='gray')
plt.title('Median Filter + Canny')

plt.show()
```
